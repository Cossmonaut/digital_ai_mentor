```python
video_ids = [
    '5wsS00QRunc', #Рушан Сахбиев, управляющий отделением "Банк Татарстан"
    '7zBkQz2BZjU', #РУШАН САХБИЕВ в Квартале Труда | Председатель Байкальского банка Сбербанка
    'jf0I2aoY3Cs', #Бизнес. По существу. Татарстан. Рушан Сахбиев - Сбербанка России Банк Татарстан
    'h-najJP4_FM', #Председатель Байкальского банка Рушан Сахбиев провел встречу с иркутскими журналистами
    'ovbP3sDZnc8', #Откровенный разговор банка и девелопера | Сбербанк & Садовое кольцо
    'fMivAUJIm2I' #Льготная ипотека от работодателя может стать конкурентным преимуществом в условиях кадрового голода
]
```

```python
for id in tqdm(video_ids):
    try:
        api_instance = YouTubeTranscriptApi()
        transcript = api_instance.fetch(video_id=id, languages=['ru', 'en'])
        full_text = " ".join([entry.text for entry in transcript])
        # Сохраняем в файл
        with open(f"data/yt_txt_rf/{id}_transcript.txt", "w", encoding="utf-8") as f:
            f.write(full_text)
    except Exception as e:
        print(f"Ошибка: {e}")
```
